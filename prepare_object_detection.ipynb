To align events and bounding boxes, you need to define the time window of the event slice. In the paper we use 1/30 second per slice.

```python
delta_T = 1/30 # the event slice length to align the bounding boxes, unit is second
```

The following code automatically loads the aligned event slices and bounding boxes, which are saved in ***aligned_data***. After that, you can add your own data-processing code.

```python
# align events and action
for index, ts in bbox_labels["ts"].items():
    start_time = (ts - delta_T) * 1e6 # convert seconds to microseconds
    end_time = ts * 1e6 # convert seconds to microseconds
    label = bbox_labels.loc[index, "bbox"]
    event_slice = event_loader.get_events(start_time, end_time)
    aligned_data = [label, event_slice] # the alinged bounding box and event sclice

    #####
    ## your own data processing
    #####
```
